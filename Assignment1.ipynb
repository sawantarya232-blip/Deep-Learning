```python
import kagglehub

path = kagglehub.dataset_download("zalando-research/fashionmnist")

print(path)
```

```python
# Import the os library to work with folders and files
import os

# Display the files available in the downloaded dataset folder
print(os.listdir(path))
```

```python
# Install TensorFlow
!pip install tensorflow

# Import TensorFlow
import tensorflow as tf

# Import Keras from TensorFlow
from tensorflow import keras

# Import libraries for data processing and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display the TensorFlow version
print("TensorFlow version:", tf.__version__)
```

```python
# Load the Fashion-MNIST training dataset
train_df = pd.read_csv(path + "/fashion-mnist_train.csv")

# Load the Fashion-MNIST testing dataset
test_df = pd.read_csv(path + "/fashion-mnist_test.csv")

# Display the first five rows of the training dataset
print(train_df.head())
```

```python
# Select all columns except the first column as training image data
X_train = train_df.iloc[:, 1:].values

# Select the first column as training labels
y_train = train_df.iloc[:, 0].values

# Select all columns except the first column as testing image data
X_test = test_df.iloc[:, 1:].values

# Select the first column as testing labels
y_test = test_df.iloc[:, 0].values
```

```python
# Normalize the training pixel values to a range between 0 and 1
X_train = X_train / 255.0

# Normalize the testing pixel values to a range between 0 and 1
X_test = X_test / 255.0
```

```python
from sklearn.model_selection import train_test_split

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)

# Display the shape of each dataset
print("Training data:", X_train.shape)
print("Validation data:", X_val.shape)
print("Testing data:", X_test.shape)
```

```python
# Display one sample image from the training dataset
plt.imshow(X_train[0].reshape(28, 28), cmap='gray')

# Display the label of the sample image
plt.title("Fashion-MNIST Sample")

# Hide the axis values
plt.axis('off')

# Show the image
plt.show()
```

```python
import matplotlib.pyplot as plt

# Display multiple sample images from the training dataset
plt.figure(figsize=(10, 5))

# Display 10 images
for i in range(10):
    plt.subplot(2, 5, i + 1)

    # Reshape the 784 pixel values into a 28x28 image
    plt.imshow(X_train[i].reshape(28, 28), cmap='gray')

    # Display the corresponding label
    plt.title("Label: " + str(y_train[i]))

    # Remove the axis values
    plt.axis('off')

# Adjust the spacing between images
plt.tight_layout()

# Display the images
plt.show()
```
